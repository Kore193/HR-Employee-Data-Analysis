### Step 02 — Data Cleaning & Transformation

```python
df['Joining_Year'] = df['Joining_Date'].dt.year
```

```python
df['Joining_Month'] = df['Joining_Date'].dt.month
```

```python
df['Joining_Month_Name'] = df['Joining_Date'].dt.month_name()
```

```python
reference_date = df['Joining_Date'].max()

df['Tenure_Years'] = (
    (reference_date - df['Joining_Date']).dt.days / 365.25
).round(1)
```

```python
bins = [0, 25, 35, 45, 55, 100]
labels = ['18-25', '26-35', '36-45', '46-55', '56+']

df['Age_Group'] = pd.cut(
    df['Age'],
    bins=bins,
    labels=labels
)
```
